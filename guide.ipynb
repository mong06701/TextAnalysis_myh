```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

if api_key:
    print("API 키를 정상적으로 불러왔습니다.")
else:
    print("API 키를 찾을 수 없습니다.")
```

```python
from dotenv import load_dotenv
from openai import OpenAI

# .env 파일 불러오기
load_dotenv()

# OpenAI 클라이언트 생성
client = OpenAI()

# API 요청
response = client.responses.create(
    model="gpt-5.6-luna",
    input="안녕하세요. 한 문장으로 답해주세요."
)

# 결과 확인
print(response.output_text)
```
